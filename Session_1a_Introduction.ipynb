## Slicing

```python
# List of sensor readings
sensor_readings = [25.5, 60.2, 101.3, 0.03, 500.1]

# Exercise 1: Get a subset of sensor readings
# Get readings from the third to the fifth sensor (indices 2, 3, and 4) using slicing and display them.
subset_readings = sensor_readings[2:5]
print("Subset of sensor readings:", subset_readings)
```

```python
# Exercise 2: Get the first N sensor readings
# Ask the user to input the number of readings they want and display them using slicing.
n = int(input("Enter the number of readings to retrieve: "))
first_n_readings = sensor_readings[:n]
print(f"First {n} sensor readings:", first_n_readings)
```

```python
# Exercise 3: Get the last N sensor readings
# Ask the user to input the number of readings they want and display them using slicing.
n = int(input("Enter the number of readings to retrieve: "))
last_n_readings = sensor_readings[-n:]
print(f"Last {n} sensor readings:", last_n_readings)
```

```python
# Exercise 4: Reverse the list of sensor readings
# Reverse the sensor_readings list using slicing and display it.
reversed_readings = sensor_readings[::-1]
print("Reversed sensor readings:", reversed_readings)
```

# Loops with FOR

```python
# Exercise 1: Iterating Through a List of Sensors
sensors = ["temperature_sensor", "humidity_sensor", "pressure_sensor", "vibration_sensor"]

for sensor in sensors:
    print("Sensor:", sensor)
```

```python
# Exercise 2: Sum of Sensor Readings
sensor_readings = [25.5, 60.2, 101.3, 0.03, 500.1]
total_reading = 0

for reading in sensor_readings:
    total_reading += reading

print("Sum of sensor readings:", total_reading)
```

# Loops with While

```python
# Exercise 1: Countdown Timer
timer = 10

while timer > 0:
    print("Countdown:", timer)
    timer -= 1

print("Blast off!")
```

```python
# Exercise 2: User Input Validation
user_input = -1

while user_input <= 0:
  user_input = float(input("Enter a positive number: "))
  if user_input <= 0:
      print("Please enter a positive number.")

print("You entered:", user_input)
```

```python
# Exercise 3: Sensor Data Retrieval
max_data_points = 5
data_points = []

while len(data_points) < max_data_points:
    # Simulate retrieving sensor data
    data_point = random.uniform(0, 100)
    data_points.append(data_point)

print("Sensor data points:", data_points)
```

```python
# Exercise 4: Data Threshold Monitoring
threshold = 100
current_temperature = 85

while current_temperature < threshold:
    # Simulate updating current temperature
    current_temperature += 5
    print("Current temperature:", current_temperature)

print("Temperature threshold crossed!")
```
